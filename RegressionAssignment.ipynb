1. Loading Dataset and Preprocessing

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

#1.Loading Dataset

calHousing = fetch_california_housing()

#2.Convert to Dataframe

df = pd.DataFrame(calHousing.data, columns=calHousing.feature_names)

df.head()

print(df.shape)

print(df.info())

print(df.describe())

print(df.isnull().sum())
df = df.fillna(df.mean())

#Standard Scalar to Standardize the feature

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)

df_scaled = pd.DataFrame(
    scaled_data,
    columns=df.columns
)

df_scaled.head()
```

3. Preprocessing steps:

a. Loaded the dataset using the function fetch_california_housing().


b. Converted the dataset into a Pandas DataFrame.


c. Checked the dataset shape, information and statistical summary.


d. Checked missing values and handled them using the mean method


e. Applied StandardScaler to standardize the features.

Preprocessing helps clean and organize the dataset. It handles missing values and brings numerical features to a same scale and data will be ready for the algorithms.

2. Regression Algorithm Implementations.

```python
df['MedianHouseVal'] = calHousing.target
df.head()

x = df.drop('MedianHouseVal', axis=1)

y = df['MedianHouseVal']

print("Input Features:", x.shape)
print("Target Variable:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42
)

print("Training Data:", x_train.shape)
print("Testing Data:", x_test.shape)

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)

x_test_scaled = scaler.transform(x_test)
```

```python
#1. Linear Regression:

from sklearn.linear_model import LinearRegression

linearReg = LinearRegression()

linearReg.fit(x_train_scaled, y_train)

linearReg_pred = linearReg.predict(x_test_scaled)

print(linearReg_pred[:5].round(2))
```

Linear Regression is used to predict continuous numerical values.In this assignment Linear Regression uses features such as median income, house age, average rooms, population, latitude and longitude to predict median house prices. It is suitable because its simple and easy to implement and helps understand the relationship between input features and house prices.

```python
#2. Decision Tree Regression

from sklearn.tree import DecisionTreeRegressor

decision_train = DecisionTreeRegressor(random_state=42)

decision_train.fit(x_train, y_train)

decision_train_pred = decision_train.predict(x_test)

print(decision_train_pred[:5].round(2))
```

Decision Tree Regression algorithm predicts continuous values by dividing the dataset into smaller groups based on conditions. In this assignment, it uses housing features to predict median house prices. It is suitable because, it can identify complex relationships between features and house prices and also does not need feature scaling.

```python
#3. Random Forest Regression

from sklearn.ensemble import RandomForestRegressor
randomF_Reg = RandomForestRegressor(n_estimators=100, random_state=42)

randomF_Reg.fit(x_train, y_train)

randomF_Reg_pred = randomF_Reg.predict(x_test)

print(randomF_Reg_pred[:5].round(2))
```

Random Forest Regression algorithm combines multiple Decision Trees to make predictions. In this assignment, 100 decision trees are used to predict median house prices. Its suitable because, it provides better prediction accuracy than a single decision tree and it reduces overfitting compared with an individual decision tree.

```python
#4.Gradient Boosting Regression

from sklearn.ensemble import GradientBoostingRegressor

gradient_Boost = GradientBoostingRegressor(random_state=42)

gradient_Boost.fit(x_train, y_train)

gradient_Boost_pred = gradient_Boost.predict(x_test)

print(gradient_Boost_pred[:5].round(2))
```

Gradient Boosting Regression algorithm builds decision trees one after another. Each new tree tries to correct the errors made by the previous trees. In this assignment, Gradient Boosting is used to predict median house prices based on housing features. Its suitable because, it can capture complex relationships in the dataset and improves predictions by correcting previous errors.

```python
#5. Support Vector Regression

from sklearn.svm import SVR

support_Vecreg = SVR(kernel='rbf')

support_Vecreg.fit(x_train_scaled, y_train)

support_Vecreg_pred = support_Vecreg.predict(x_test_scaled)

print(support_Vecreg_pred[:5].round(2))
```

Support Vector Regression (SVR) algorithm is used to predict continuous numerical values. It tries to find a function that predicts values while keeping errors within a specified margin as much as possible. Its suitable because, It can handle non-linear relationships using kernels and works well with properly scaled numerical features.

3. Model Evaluation and Comparison

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

models_Prediction = {
    'Linear Regression': linearReg_pred,
    'Decision Tree': decision_train_pred,
    'Random Forest': randomF_Reg_pred,
    'Gradient Boosting': gradient_Boost_pred,
    'SVR': support_Vecreg_pred
}

results = []

for name, prediction in models_Prediction.items():
    mse = mean_squared_error(y_test, prediction)
    mae = mean_absolute_error(y_test, prediction)
    r2 = r2_score(y_test, prediction)

    results.append([name, mse, mae, r2])

    results_df = pd.DataFrame(
    results,
    columns=['Algorithm', 'MSE', 'MAE', 'R2 Score']
)

print(results_df)
```

The five regression algorithms were evaluated using Mean Squared Error (MSE), Mean Absolute Error (MAE), and R² Score. Among all five algorithms, Random Forest Regressor performed the best, with the lowest MSE of 0.255591, lowest MAE of 0.327816, and highest R² Score of 0.804953. Linear Regression performed the worst, with the highest MSE of 0.555892, highest MAE of 0.533200, and lowest R² Score of 0.575788.

Therefore, Random Forest Regressor is the most suitable algorithm among the five tested models.
